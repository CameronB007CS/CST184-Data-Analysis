```python
import pandas as pd
```

```python
df = pd.read_csv("energy_cleaned.csv")

print(df.head())

print(df.info())

print(df.describe())
```

```python
high_renew = df[df["%Renewable"] > 50]
print("Countries with >50% renewable energy:", len(high_renew))
```

```python
high_renew_60 = df[df["%Renewable"] > 60]
high_renew_70 = df[df["%Renewable"] > 70]
print("Countries with >60% renewable energy:", len(high_renew_60))
print("Countries with >70% renewable energy:", len(high_renew_70))
```

```python
import matplotlib.pyplot as plt
```

```python
region_avg = df.groupby("Region")["%Renewable"].mean()
print(region_avg)
```

```python
region_avg.plot(kind="bar")
```

```python
import matplotlib.pyplot as plt
```

```python
df["%Renewable"].plot(kind="hist", bins=8, edgecolor="black")
plt.title("Histogram of Renewable Energy Use")
plt.xlabel("% Renewable")
plt.grid(True)
plt.show()
```

```python
df["%Renewable"].plot(kind="hist", bins=5, edgecolor="black")
plt.title("Histogram of Renewable Energy Use (5 Bins)")
plt.xlabel("% Renewable")
plt.grid(True)
plt.show()
```

```python
df["%Renewable"].plot(kind="density", linewidth=2)
plt.title("Density Plot of Renewable %")
plt.xlabel("% Renewable")
plt.grid(True)
plt.show()
```

```python
high_renew = df[df["%Renewable"] > 50]
high_renew["%Renewable"].plot(kind="density", linewidth=2)
plt.title("Density Plot of >50% Renewable")
plt.xlabel("% Renewable")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
```

```python
for region, group in df.groupby("Region"):
    if len(group) >= 2:
        group["%Renewable"].plot(kind="density", label=region)

plt.title("Renewable Energy by Region (KDE)")
plt.xlabel("% Renewable")
plt.legend()
plt.grid(True)
plt.show()
```

```python
for region, group in df.groupby("Region"):
    group["%Renewable"].plot(kind="hist", alpha=0.5, label=region, bins=5)

plt.title("Renewable Energy Histograms by Region")
plt.xlabel("% Renewable")
plt.legend()
plt.grid(True)
plt.show()
```
